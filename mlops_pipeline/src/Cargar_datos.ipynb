# Carga de datos

Simula el proceso de materializacion de los datos que en un entorno
productivo entregaria el DWH o el Datalake. La fuente es una tabla de
BigQuery en Google Cloud; el proyecto, el dataset y la tabla se leen de
`mlops_pipeline/config.json` (seccion `datos`).

**Fuente:** `retail-apps-datos-dev-01.creditos.dbcreditos` (creditos de consumo, Colombia)
**Autenticacion:** credenciales predeterminadas de aplicacion (`gcloud auth application-default login`), sin llaves de cuenta de servicio
**Objetivo:** `Pago_atiempo` (1 = obligacion pagada en su totalidad)

```python
import logging
import sys
from pathlib import Path

# Ubica la raiz del repositorio sin importar desde donde se abra el notebook
RAIZ = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "mlops_pipeline").exists())
sys.path.insert(0, str(RAIZ / "mlops_pipeline" / "src"))

from ft_engineering import cargar_config, cargar_datos

logging.basicConfig(level=logging.INFO, format="%(levelname)s | %(message)s", force=True)

config = cargar_config()
datos = cargar_datos(config)
datos.head()
```

```python
print(f"Registros: {datos.shape[0]:,}")
print(f"Columnas:  {datos.shape[1]}")
print(f"Duplicados: {datos.duplicated().sum()}")
print(f"\nDistribucion del objetivo:")
print(datos[config["datos"]["columna_objetivo"]].value_counts(normalize=True).round(4))
print(f"\nTipos de dato:")
print(datos.dtypes.value_counts())
```
